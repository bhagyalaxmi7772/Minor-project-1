Chat Parser


```python
with open("/content/DADS Minor PROJECT dataset.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total lines:", len(lines))
print(lines[:5])
```

```python
messages = []

for line in lines:
    line = line.strip()

    if " - " not in line:
        continue

    date_time, content = line.split(" - ", 1)

    if ": " in content:
        sender, message = content.split(": ", 1)
    else:
        sender = "System"
        message = content

    if ", " in date_time:
        date, time = date_time.split(", ", 1)
    else:
        continue

    messages.append({
        "date": date,
        "time": time,
        "sender": sender,
        "message": message
    })

print("Parsed messages:", len(messages))
print(messages[:5])
```

Group Overview

```python
total_messages = len(messages)

system_messages = 0
participants = set()

for msg in messages:
    if msg["sender"] == "System":
        system_messages += 1
    else:
        participants.add(msg["sender"])

user_messages = total_messages - system_messages

print("GROUP OVERVIEW")
print("---------------------------")
print("Total Messages:", total_messages)
print("System Messages:", system_messages)
print("User Messages:", user_messages)
print("Total Participants:", len(participants))
print("Participants:")
print(participants)
```

Messages Per Participant

```python
message_count = {}

for msg in messages:
    sender = msg["sender"]

    if sender != "System":
        if sender in message_count:
            message_count[sender] += 1
        else:
            message_count[sender] = 1

print("MESSAGES PER PERSON")
print("-------------------------")

for sender, count in message_count.items():
    print(sender, ":", count)
```

Messages Per Day

```python
day_count = {}

for msg in messages:
    date = msg["date"]

    if date in day_count:
        day_count[date] += 1
    else:
        day_count[date] = 1

print("MESSAGES PER DAY")
print("---------------------------")

for date, count in day_count.items():
    print(date, ":", count)
```

Messages Per Hour

```python
hour_count = {}

for msg in messages:
    hour = msg["time"].split(":")[0]

    if hour in hour_count:
        hour_count[hour] += 1
    else:
        hour_count[hour] = 1

print("MESSAGES PER HOUR")
print("---------------------------")

for hour, count in sorted(hour_count.items()):
    print(hour + ":00 -", count)
```

Top Words

```python
word_count = {}

for msg in messages:
    if msg["sender"] == "System":
        continue

    words = msg["message"].lower().split()

    for word in words:
        word = word.strip(".,!?()[]{}\"'")

        if word == "":
            continue

        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

sorted_words = sorted(word_count.items(), key=lambda x: x[1], reverse=True)

print("TOP 10 MOST USED WORDS")
print("---------------------------")

for word, count in sorted_words[:10]:
    print(word, ":", count)
```

Activity Heatmap

```python
import numpy as np

heatmap = np.zeros(24, dtype=int)

for msg in messages:
    hour = int(msg["time"].split(":")[0])
    heatmap[hour] += 1

print("ACTIVITY HEATMAP")
print("---------------------------")

for i in range(24):
    print(f"{i:02d}:00 - {heatmap[i]} messages")
```

Response Time

```python
from datetime import datetime

response_times = []

for i in range(1, len(messages)):
    previous = messages[i - 1]
    current = messages[i]

    if previous["sender"] != current["sender"]:

        t1 = datetime.strptime(previous["date"] + " " + previous["time"], "%d/%m/%y %H:%M")
        t2 = datetime.strptime(current["date"] + " " + current["time"], "%d/%m/%y %H:%M")

        diff = (t2 - t1).total_seconds() / 60

        if diff >= 0:
            response_times.append(diff)

if response_times:
    avg_response = sum(response_times) / len(response_times)

    print("AVERAGE RESPONSE TIME")
    print("---------------------------")
    print(round(avg_response, 2), "minutes")
```

Longest Silent Streak

```python
longest_gap = 0

for i in range(1, len(messages)):
    t1 = datetime.strptime(messages[i-1]["date"] + " " + messages[i-1]["time"], "%d/%m/%y %H:%M")
    t2 = datetime.strptime(messages[i]["date"] + " " + messages[i]["time"], "%d/%m/%y %H:%M")

    gap = (t2 - t1).total_seconds() / 3600

    if gap > longest_gap:
        longest_gap = gap

print("LONGEST SILENT STREAK")
print("---------------------------")
print(round(longest_gap,2), "hours")
```
